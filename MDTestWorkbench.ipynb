## Start here

**The easiest way to run MDTest is to use the START launcher in the repository folder.** It prepares the Python environment and opens this notebook in JupyterLab automatically. You do **not** need to link the `src` folder, set `PYTHONPATH`, or run `pip` yourself.

Once the Workbench appears below, use **Load columns** to confirm the data and observable, choose **Auto Compare** for a first comparison, and select **Run analysis**. The scientific tests are identical in the GUI and Python API.

If you opened this notebook manually and get `No module named mdtest`, close it and launch through `START_MDTEST_WINDOWS.bat` (Windows), `START_MDTEST_MAC.command` (macOS), or `bash START_MDTEST_LINUX.sh` (Linux). For first-time instructions, see `START_HERE.html` or `docs/START_HERE.md`.

## Launch the interactive workspace

The next cell opens the product-style Workbench. The interface keeps data selection, analysis choices, result interpretation, and export visibly separated while every numerical run is dispatched through the public API.

## What to look for after a run

The integrated result surface is designed to answer four practical questions in order:

1. **Where does MDTest first accept the configured stationarity screens?**
2. **Where does ESS/IACT place the correlation-efficiency optimum?**
3. **What reporting block size passes the complete four-test diagnostic protocol?**
4. **Can the result be reproduced from the recorded source hash, settings, software versions, and exported structured result?**

Diagnostic figures are embedded inside the same dashboard rather than appearing as unrelated notebook outputs. A diagnostic failure is a scientific result, not a software error.


## Reproducible non-widget workflow

The GUI is optional. The cell below runs the same controller/API path directly and displays the same integrated dashboard. This is useful for scripted notebooks, teaching material, and audit/reproducibility checks.


```python
from pathlib import Path
from mdtest.workbench import WorkbenchRequest, run_request
from mdtest.presentation import display_result

example = Path("examples/data/synthetic_variance_relaxation_independent.csv")
if example.exists():
    request = WorkbenchRequest(
        file=str(example),
        col=1,
        mode="auto-compare",
        m_trend_grid=(2, 5, 10, 20, 40),
        m_corr_grid=(10, 20, 40, 80, 160, 320),
    )
    result = run_request(request)
    display_result(result, level="guided", include_figures=True)
else:
    print("Example data were not found relative to the current working directory.")
```

## Export a portable report

A successful current result can be exported as **JSON + guided text + a self-contained HTML report + PNG diagnostics**. The HTML embeds its figures, so it can be moved or shared without sibling image files.


```python
from mdtest.presentation import export_guided_report

if "result" in globals():
    exported = export_guided_report(result, "results/workbench_notebook", prefix="mdtest_notebook")
    exported
else:
    print("Run the reproducible example above, or export directly from the interactive Workbench.")
```
